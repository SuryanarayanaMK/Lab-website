The dataset contains **1,797 images** of handwritten digits 0–9. Each image is an **8 × 8 grid** of pixel intensities from **0 (background)** to **16 (dark stroke)**, so each image can be flattened into **64 numerical features**.

Each row is one image (not one pixel).

| Object | Description |
|---|---|
| `digits.images[i]` | Image `i` as an 8 × 8 array |
| `digits.data[i]` | The same image flattened into 64 pixel values |
| `digits.target[i]` | Correct label of that image (0–9) |
| `pixel_r_c` (in `df`) | Intensity of the pixel at row `r`, column `c` |
| `digit` (in `df`) | Label column |

This notebook only loads the data and makes a few basic plots. The analysis is up to you.

> The data ship with scikit-learn, so no download is needed.

## 1. Load the data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

digits = load_digits()
images = digits.images   # (n_images, 8, 8)
X = digits.data          # (n_images, 64)
y = digits.target        # labels 0-9

pixel_columns = [f'pixel_{r}_{c}' for r in range(8) for c in range(8)]
df = pd.DataFrame(X.astype(int), columns=pixel_columns)
df['digit'] = y
print(df.shape)
df.head()
```

```python
df.describe().T
```

## 2. A first look

```python
fig, axs = plt.subplots(2, 5, figsize=(11, 5), constrained_layout=True)
for digit, ax in enumerate(axs.flat):
    idx = np.flatnonzero(y == digit)[0]
    ax.imshow(images[idx], cmap='gray_r', vmin=0, vmax=16, interpolation='nearest')
    ax.set(title=f'Digit {digit}', xticks=[], yticks=[])
fig.suptitle('One example image per digit')
plt.show()
```
